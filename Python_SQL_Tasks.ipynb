## Task 2: Check palindrome

**Prompt:** Write a program to check whether a string is a palindrome, ignoring letter case.

```python
text = "Madam"
normalized = text.lower()
print("Is palindrome:", normalized == normalized[::-1])
```

## Task 3: Count vowels

**Prompt:** Count the vowels in a string.

```python
text = "Python Programming"
vowels = set("aeiou")
count = sum(1 for ch in text.lower() if ch in vowels)
print("Vowel count:", count)
```

## Task 4: Character frequency

**Prompt:** Count how many times each character appears in a string.

```python
text = "banana"
frequency = {}
for ch in text:
    frequency[ch] = frequency.get(ch, 0) + 1
print(frequency)
```

## Task 5: Factorial

**Prompt:** Calculate the factorial of a non-negative integer using a loop.

```python
n = 5
factorial = 1
for number in range(1, n + 1):
    factorial *= number
print(f"{n}! =", factorial)
```

## Task 6: Prime number

**Prompt:** Check whether a number is prime.

```python
n = 29
is_prime = n >= 2
for divisor in range(2, int(n ** 0.5) + 1):
    if n % divisor == 0:
        is_prime = False
        break
print(f"{n} is prime:", is_prime)
```

## Task 7: Second-largest distinct number

## Task 10: Rotate a list right

**Prompt:** Rotate a list to the right by k positions.

```python
numbers = [1, 2, 3, 4, 5]
k = 2
k %= len(numbers)
rotated = numbers[-k:] + numbers[:-k] if k else numbers[:]
print(rotated)
```

# SQL Interview Tasks

SQL examples run inside an in-memory SQLite database. The setup cell creates sample `Employees` and `Departments` tables. Run tasks in order because later tasks use the same database.

## SQL setup

**Prompt:** Create sample employee and department tables and insert data for interview practice.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
cur = conn.cursor()

cur.executescript("""
DROP TABLE IF EXISTS Employees;
DROP TABLE IF EXISTS Departments;

CREATE TABLE Departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT NOT NULL
);

CREATE TABLE Employees (
    employee_id INTEGER PRIMARY KEY,
    employee_name TEXT NOT NULL,
    department_id INTEGER,
    salary INTEGER,
    manager_id INTEGER,
    FOREIGN KEY (department_id) REFERENCES Departments(department_id)
);

INSERT INTO Departments VALUES
(1, 'Engineering'), (2, 'HR'), (3, 'Sales');

INSERT INTO Employees VALUES
(101, 'Anu', 1, 60000, NULL),
(102, 'Ravi', 1, 75000, 101),
(103, 'Kiran', 2, 50000, 101),
(104, 'Meena', 3, 55000, 102),
(105, 'Arjun', 1, 75000, 102),
(106, 'Divya', 2, 45000, 103);
""")
conn.commit()

def show_query(query):
    display(pd.read_sql_query(query, conn))

print("Tables created and sample data inserted.")
display(pd.read_sql_query("SELECT * FROM Employees ORDER BY employee_id", conn))
```

## Task 23: SELECT and WHERE

**Prompt:** Select employees whose salary is greater than 55000.

```python
show_query("""
SELECT employee_name, salary
FROM Employees
WHERE salary > 55000
ORDER BY salary DESC;
""")
```

## Task 24: ORDER BY and LIMIT

**Prompt:** Display the three highest-paid employees.

```python
show_query("""
SELECT employee_name, salary
FROM Employees
ORDER BY salary DESC
LIMIT 3;
""")
```

## Task 25: Aggregate functions

**Prompt:** Find the number of employees, average salary, minimum salary, and maximum salary.

```python
show_query("""
SELECT COUNT(*) AS employee_count,
       ROUND(AVG(salary), 2) AS average_salary,
       MIN(salary) AS minimum_salary,
       MAX(salary) AS maximum_salary
FROM Employees;
""")
```

## Task 26: GROUP BY and HAVING

**Prompt:** Show departments whose average salary is greater than 55000.

```python
show_query("""
SELECT d.department_name, ROUND(AVG(e.salary), 2) AS average_salary
FROM Employees e
JOIN Departments d ON e.department_id = d.department_id
GROUP BY d.department_name
HAVING AVG(e.salary) > 55000;
""")
```

## Task 27: INNER JOIN

**Prompt:** Display each employee with their department name using an INNER JOIN.

```python
show_query("""
SELECT e.employee_name, d.department_name
FROM Employees e
INNER JOIN Departments d ON e.department_id = d.department_id
ORDER BY e.employee_id;
""")
```

## Task 28: LEFT JOIN

**Prompt:** Display all departments and any employees in each department, including departments with no employees.

```python
show_query("""
SELECT d.department_name, e.employee_name
FROM Departments d
LEFT JOIN Employees e ON d.department_id = e.department_id
ORDER BY d.department_id, e.employee_id;
""")
```

## Task 29: UNION

**Prompt:** Combine employee names with department names using UNION, removing duplicates.

```python
show_query("""
SELECT employee_name AS name FROM Employees
UNION
SELECT department_name AS name FROM Departments
ORDER BY name;
""")
```

## Task 30: Find duplicate values

**Prompt:** Find salary values shared by more than one employee.

```python
show_query("""
SELECT salary, COUNT(*) AS number_of_employees
FROM Employees
GROUP BY salary
HAVING COUNT(*) > 1;
""")
```

## Task 31: Subquery

**Prompt:** Find employees earning more than the overall average salary.

```python
show_query("""
SELECT employee_name, salary
FROM Employees
WHERE salary > (SELECT AVG(salary) FROM Employees)
ORDER BY salary DESC;
""")
```

## Task 32: LEAD and LAG window functions

**Prompt:** Show each employee's salary along with the previous and next salary when employees are ordered by employee ID.

```python
show_query("""
SELECT employee_name, salary,
       LAG(salary) OVER (ORDER BY employee_id) AS previous_salary,
       LEAD(salary) OVER (ORDER BY employee_id) AS next_salary
FROM Employees
ORDER BY employee_id;
""")
```

## Task 35: TRUNCATE equivalent in SQLite

**Prompt:** Empty a practice table. SQLite does not support TRUNCATE TABLE, so use DELETE without a WHERE clause.

```python
cur.executescript("""
DROP TABLE IF EXISTS TempRecords;
CREATE TABLE TempRecords (record_id INTEGER, label TEXT);
INSERT INTO TempRecords VALUES (1, 'A'), (2, 'B'), (3, 'C');
DELETE FROM TempRecords;
""")
show_query("SELECT COUNT(*) AS rows_after_emptying FROM TempRecords")
```

## Task 37: CASE expression

**Prompt:** Label employees as High salary if salary is at least 70000, otherwise Standard salary.

```python
show_query("""
SELECT employee_name, salary,
       CASE WHEN salary >= 70000 THEN 'High salary'
            ELSE 'Standard salary'
       END AS salary_band
FROM Employees
ORDER BY employee_id;
""")
```
